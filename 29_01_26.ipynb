```python
class Node:
    def __init__(self, data,next=None):
        self.data = data
        self.next = next

class LinkedList:
    def __init__(self):
        self.head = None
    def traversal(self):
      temp = self.head
      while (temp!=None):
        print(temp.data,end="->")
        temp=temp.next
      print()
    def insertbegin(self,data):
      new_node = Node(data)
      if self.head==None:
        self.head=new_node
      else:
        new_node.next = self.head
        self.head = new_node

ll=LinkedList()
ll.insertbegin(70)
ll.insertbegin(60)
ll.insertbegin(50)
ll.traversal()
```
